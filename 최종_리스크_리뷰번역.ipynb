# 0️⃣ 데이터 준비

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
import seaborn as sns
import copy
from datetime import datetime, timedelta
import random
import warnings
from transformers import pipeline
from tqdm import tqdm

sns.set(font="Malgun Gothic", rc={"axes.unicode_minus": False})  # 마이너스 값 깨지는 거 방지 & 그래프 디자인 배경 회색, 그리드
pd.set_option('display.max_columns', None)  # 전체 출력 O

order_reviews = pd.read_csv('olist_order_reviews_dataset.csv')
```

```python
review_trans = order_reviews[['review_id', 'order_id', 'review_score', 'review_comment_title',
       'review_comment_message', 'review_creation_date',
       'review_answer_timestamp']].copy()
```

# 1️⃣ googletrans 사용

```python
from googletrans import Translator

translator = Translator()

# 에러 난 text 저장
err_text = []

def translate_to_english(text):
    # NA이거나 텍스트가 없으면 빈 문자열 반환
    if pd.isna(text) or text == '':
        return ''
    
    try:
        result = translator.translate(text, src='pt', dest='en')
        return result.text
    except Exception as e:
        # 에러 발생 시 원래 텍스트 반환하거나 에러 메시지 출력
        print(f"Error converting: {text}, Error: {e}")
        err_text.append(text)
        return text 
```

```python
print("전체 데이터 번역 시작...")
tqdm.pandas()
review_trans['review_comment_translated'] = review_trans['review_comment_message'].progress_apply(translate_to_english)

# 5. 결과 확인
print(review_trans[['review_comment_message', 'review_comment_translated']].head())
```

```python
eq_cnt = 0
idx = []
for i in range(len(review_trans)):
    if review_trans['review_comment_message'].iloc[i] == review_trans['review_comment_translated'].iloc[i]:
        eq_cnt+=1
        idx.append(i)
print(eq_cnt) 
print(idx)

# equal_count(review_trans)
er_reviews = review_trans[['review_comment_message','review_comment_translated']].iloc[idx]
er_reviews

print("에러 데이터 번역 시작...")
tqdm.pandas()
er_reviews['review_comment_translated'] = er_reviews['review_comment_message'].progress_apply(translate_to_english)
```
